```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger le fichier CSV
df = pd.read_csv("zara_apres_traitement.csv")
```

```python
# 1. Histogramme des prix
# ============================
plt.figure()
plt.hist(df['price'].dropna())
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Histogram of Prices')
plt.show()
```

```python
# 2. Volume moyen des ventes par catégorie
# =============================================
plt.figure()
df.groupby('Product Category')['Sales Volume'].mean().plot(kind='bar')
plt.xlabel('Product Category')
plt.ylabel('Average Sales Volume')
plt.title('Average Sales Volume by Product Category')
plt.show()
```

```python
# 3. Nombre de produits par section (MAN / WOMAN)
# =============================================
plt.figure()
df['section'].value_counts().plot(kind='bar')
plt.xlabel('Section')
plt.ylabel('Product Count')
plt.title('Number of Products per Section')
plt.show()
```

```python
# 4. Histogramme des Sales Volume
# ===============================
plt.figure()
plt.hist(df['Sales Volume'].dropna())
plt.xlabel('Sales Volume')
plt.ylabel('Frequency')
plt.title('Histogram of Sales Volume')
plt.show()
```
